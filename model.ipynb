<h3>Popularity Based Recommendation System</h3>

```python
ratings_new=ratings.merge(books,on='ISBN')
ratings_new.head()
```

```python
num_rating_df = ratings_new.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'Num_ratings'},inplace=True)
num_rating_df
```

```python
ratings_new['Book-Rating'] = pd.to_numeric(ratings_new['Book-Rating'], errors='coerce')
avg_rating_df = ratings_new.groupby('Book-Title')['Book-Rating'].mean().reset_index()
avg_rating_df.rename(columns={'Book-Rating': 'Avg_rating'}, inplace=True)
avg_rating_df
```

```python
new_df=num_rating_df.merge(avg_rating_df,on='Book-Title')
new_df.sample(30)
```

```python
final_df=new_df[new_df['Num_ratings']>=250].sort_values('Avg_rating',ascending=False).head(50)
final_df
```

```python
final_df=final_df.merge(books,on='Book-Title').drop_duplicates('Book-Title')
final_df.head()
```

```python
final_df=final_df[['Book-Title','Book-Author','Image-URL-M','Num_ratings','Avg_rating']]
final_df
```

<h3>Collaborative Filtering Based Recommender System</h3>

```python
x=ratings_new.groupby('User-ID').count()['Book-Rating']>200
valid=x[x].index
valid
```

```python
filtered_df=ratings_new[ratings_new['User-ID'].isin(valid)]
filtered_df
```

```python
num_rating_df2 = filtered_df.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df2.rename(columns={'Book-Rating':'Num_ratings'},inplace=True)
num_rating_df2
```

```python
df=num_rating_df2[num_rating_df2['Num_ratings']>=50]
df
```

```python
finalized_df=filtered_df.merge(df,on='Book-Title')
finalized_df.head()
```

```python
finalized_df.drop(columns=['Num_ratings'],inplace=True)
finalized_df.drop_duplicates()
finalized_df.shape
```

```python
pt = finalized_df.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt.fillna(0,inplace=True)
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_scores = cosine_similarity(pt)
similarity_scores.shape
```

```python
def recommend(book_name):
    # index fetch
    index = np.where(pt.index==book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[1:5]
    
    for i in similar_items:
        print(pt.index[i[0]])
```

```python
recommend('1984')
```
